```python
def max_tup(tup):
    return max(tup)
def min_tup(tup):
    return min(tup)
def sum_tup(tup):
    return sum(tup)
print(f"maximum of tuple: {max_tup(tup)}")
print(f"minimum of tuple: {min_tup(tup)}")
print(f"sumation of tup: {sum_tup(tup)}")
```
