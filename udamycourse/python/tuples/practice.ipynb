```python
numbers=(1,2,3,4,5,6)
```

```python
mixed_tuple=(1,"hello",3.14,True)
```

```python
#acessing elemnts in tuple
print(numbers[0])
print(numbers[-1])
print(numbers[2:])
print(numbers[::-1])
```

```python
#tuple operation
concatination=numbers + mixed_tuple
```

```python
concatination
```

```python
mixed_tuple*3
```

```python
numbers * 3
```

```python
for i in numbers:
    print(i*3)
```

```python
#immutable nature of tuples
lst=[1,2,3,4,5]
print(lst)
lst[1]="krish"
print(lst)

numbers[2]="krish"

```

```python
#tuple method
print(numbers.count(1)) #it means how many time specific numbers occurs in tuple
print(numbers.index(4)) #given number return index 
```

```python
#packing and unpacking
#packing
packed_tuple=1,"hello",3.1
print(packed_tuple)
```

```python
#unpacking
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
#unpacking with *
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
#nested tuples
tup1=((1,2,3,4),(5,6,7,8),(1,"hello",3.4,True))
print(tup1[2][1:])
print(tup1[2][1])
```

```python
for i in tup1:
    for j in i:
        print(j,end=" ")
    print()
```
